```python
import GEOparse

gse_36059 = GEOparse.get_GEO(geo="GSE36059", destdir=RAW_DIR)
print("Samples in GSE36059:", len(gse_36059.gsms))
print("Platforms:", list(gse_36059.gpls.keys()))
```

```python
for i, (gsm_name, gsm) in enumerate(gse_36059.gsms.items()):
    if i >= 8:
        break
    print(gsm_name, "->", gsm.metadata.get("characteristics_ch1", []))
```

```python
import pandas as pd
import re

def extract_characteristics(gsm):
    """Turn a GSM's characteristics_ch1 list into a dict of {field: value}."""
    out = {}
    for entry in gsm.metadata.get("characteristics_ch1", []):
        if ":" in entry:
            key, val = entry.split(":", 1)
            out[key.strip()] = val.strip()
    return out

rows = []
for gsm_name, gsm in gse_36059.gsms.items():
    chars = extract_characteristics(gsm)
    chars["sample_id"] = gsm_name
    rows.append(chars)

meta_36059 = pd.DataFrame(rows).set_index("sample_id")
print(meta_36059.columns.tolist())
meta_36059.head()
```

```python
DIAGNOSIS_COL = "diagnosis (tcmr,abmr,mixed,non-rejecting)"

LABEL_MAP_36059 = {
    "non-rejecting": "NR",
    "tcmr": "TCMR",
    "abmr": "ABMR",
    "mixed": "Mixed",
}

meta_36059["rejection_class"] = (
    meta_36059[DIAGNOSIS_COL].str.lower().str.strip().map(LABEL_MAP_36059)
)

print(meta_36059["rejection_class"].value_counts(dropna=False))
```

```python
nan_rows = meta_36059[meta_36059["rejection_class"].isna()]
print(nan_rows[DIAGNOSIS_COL].unique())
nan_rows[[DIAGNOSIS_COL]]
```

```python
meta_36059_clean = meta_36059.dropna(subset=["rejection_class"]).copy()

print(f"Dropped {len(meta_36059) - len(meta_36059_clean)} non-transplant/unlabeled samples")
print(f"Final GSE36059 labeled samples: {len(meta_36059_clean)}")
print(meta_36059_clean["rejection_class"].value_counts())
```

```python
expr_frames = []
for gsm_name in meta_36059_clean.index:
    gsm = gse_36059.gsms[gsm_name]
    if gsm.table is None or gsm.table.empty:
        continue
    series = gsm.table.set_index("ID_REF")["VALUE"]
    series.name = gsm_name
    expr_frames.append(series)

expr_36059 = pd.concat(expr_frames, axis=1)
print("Expression matrix shape (probes x samples):", expr_36059.shape)
expr_36059.head()
```

```python
platform_ids = list(gse_36059.gpls.keys())
print("Platform(s):", platform_ids)

gpl = gse_36059.gpls[platform_ids[0]]
anno = gpl.table
print("Annotation columns:", anno.columns.tolist())
anno.head()
```

```python
probe_to_gene = anno.set_index("ID")["Gene Symbol"]

expr_36059_genes = expr_36059.join(probe_to_gene.rename("gene_symbol"), how="left")

# Drop probes with no gene symbol (control probes, unannotated, etc.)
before = len(expr_36059_genes)
expr_36059_genes = expr_36059_genes.dropna(subset=["gene_symbol"])
expr_36059_genes = expr_36059_genes[expr_36059_genes["gene_symbol"].str.strip() != ""]
print(f"Dropped {before - len(expr_36059_genes)} unannotated/control probes")

# Multiple probes can map to the same gene (common on Affy arrays) —
# collapse by taking the mean expression across probes per gene
expr_36059_final = expr_36059_genes.groupby("gene_symbol").mean(numeric_only=True)

print("Gene-level expression matrix shape (genes x samples):", expr_36059_final.shape)
expr_36059_final.head()
```

```python
common_samples = expr_36059_final.columns.intersection(meta_36059_clean.index)
print(f"Samples in expression matrix: {expr_36059_final.shape[1]}")
print(f"Samples in metadata: {len(meta_36059_clean)}")
print(f"Common samples: {len(common_samples)}")

expr_36059_final = expr_36059_final[common_samples]
meta_36059_clean = meta_36059_clean.loc[common_samples]

assert (expr_36059_final.columns == meta_36059_clean.index).all()
print("Aligned successfully:", expr_36059_final.shape[1], "samples")
```

```python
from combat.pycombat import pycombat

# pycombat expects: expression dataframe (genes x samples), batch labels aligned to columns
batch_labels = combined_meta.loc[combined_expr.columns, "study"]

combined_expr_corrected = pycombat(combined_expr, batch_labels)

print("Batch-corrected expression matrix shape:", combined_expr_corrected.shape)
print("Value range after correction:", combined_expr_corrected.values.min(), "-", combined_expr_corrected.values.max())
```

```python
from sklearn.model_selection import train_test_split

RANDOM_SEED = 42
TRAIN_FRAC, VAL_FRAC, TEST_FRAC = 0.70, 0.15, 0.15

train_ids, temp_ids = train_test_split(
    combined_meta.index,
    test_size=(1 - TRAIN_FRAC),
    stratify=combined_meta["rejection_class"],
    random_state=RANDOM_SEED,
)

val_size = VAL_FRAC / (VAL_FRAC + TEST_FRAC)
val_ids, test_ids = train_test_split(
    combined_meta.loc[temp_ids].index,
    test_size=(1 - val_size),
    stratify=combined_meta.loc[temp_ids, "rejection_class"],
    random_state=RANDOM_SEED,
)

combined_meta["split"] = "train"
combined_meta.loc[val_ids, "split"] = "val"
combined_meta.loc[test_ids, "split"] = "test"

print(combined_meta.groupby(["split", "rejection_class"]).size().unstack(fill_value=0))
```

```python
import numpy as np

RANDOM_SEED = 42
TRAIN_FRAC, VAL_FRAC, TEST_FRAC = 0.70, 0.15, 0.15

rng = np.random.default_rng(RANDOM_SEED)

combined_meta["split"] = None

for cls, group in combined_meta.groupby("rejection_class"):
    ids = group.index.to_numpy().copy()
    rng.shuffle(ids)

    n = len(ids)
    n_train = int(round(n * TRAIN_FRAC))
    n_val = int(round(n * VAL_FRAC))

    train_ids = ids[:n_train]
    val_ids = ids[n_train:n_train + n_val]
    test_ids = ids[n_train + n_val:]

    combined_meta.loc[train_ids, "split"] = "train"
    combined_meta.loc[val_ids, "split"] = "val"
    combined_meta.loc[test_ids, "split"] = "test"

print(combined_meta.groupby(["split", "rejection_class"]).size().unstack(fill_value=0))
```

```python
complete_rows = clinical_98320_final.dropna()
print(f"Samples with fully complete clinical data: {len(complete_rows)} out of {len(clinical_98320_final)}")

# Check class balance is preserved after dropping
complete_labels = meta_98320_clinical.loc[complete_rows.index, "rejection_class"]
print("\nClass distribution if we drop incomplete rows:")
print(complete_labels.value_counts())
```

```python
CLINICAL_COLS_MEDIAN = clinical_98320_final.median()

clinical_98320_imputed = clinical_98320_final.copy()
was_imputed = clinical_98320_final.isna().any(axis=1).astype(int)

for col in CLINICAL_COLS:
    clinical_98320_imputed[col] = clinical_98320_imputed[col].fillna(CLINICAL_COLS_MEDIAN[col])

clinical_98320_imputed["was_imputed"] = was_imputed

print("Imputed clinical table shape:", clinical_98320_imputed.shape)
print(f"Rows with at least one imputed value: {was_imputed.sum()} out of {len(was_imputed)}")
print(clinical_98320_imputed.isna().sum().sum(), "remaining NaNs (should be 0)")
```

```python
multimodal_dataset = expr_98320_clinical_final.T  # samples x genes (732 samples, all of them now)
multimodal_dataset = multimodal_dataset.join(clinical_98320_imputed)  # clinical cols + was_imputed flag
multimodal_dataset = multimodal_dataset.join(meta_98320_clinical[["rejection_class"]])

print("Multimodal dataset shape (samples x [genes + clinical + was_imputed + label]):", multimodal_dataset.shape)
print("\nClass distribution:")
print(multimodal_dataset["rejection_class"].value_counts())
print("\nImputed vs. complete breakdown:")
print(multimodal_dataset["was_imputed"].value_counts())

# Save
multimodal_dataset.to_parquet(os.path.join(PROCESSED_DIR, "GSE98320_multimodal.parquet"))
multimodal_dataset.to_csv(os.path.join(PROCESSED_DIR, "GSE98320_multimodal.csv"))

print("\nSaved GSE98320_multimodal.parquet and .csv")
```
